## Question 1

Create a ranking of students based on score (highest first).

```python
query = '''
SELECT
    student_id,
    name,
    score,
    RANK() OVER (ORDER BY score DESC) AS score_rank
FROM Student_Performance;
'''
print(query)
```

**Answer:** This uses `RANK()` to assign rank 1 to the highest score. Students with the same score receive the same rank.

## Question 2

Show each student's score and the previous student’s score (based on score order).

```python
query = '''
SELECT
    student_id,
    name,
    score,
    LAG(score) OVER (ORDER BY score DESC) AS previous_score
FROM Student_Performance;
'''
print(query)
```

**Answer:** `LAG(score)` returns the score from the previous row when students are ordered by score from highest to lowest.

## Question 3

Convert all student names to uppercase and extract the month name from join_date.

```python
query = '''
SELECT
    student_id,
    UPPER(name) AS student_name_upper,
    MONTHNAME(join_date) AS join_month
FROM Student_Performance;
'''
print(query)
```

**Answer:** `UPPER()` converts names to uppercase, while `MONTHNAME()` extracts the month name from `join_date`.

## Question 4

Show each student's name and the next student’s attendance (ordered by attendance).

```python
query = '''
SELECT
    name,
    attendance,
    LEAD(attendance) OVER (ORDER BY attendance) AS next_student_attendance
FROM Student_Performance;
'''
print(query)
```

**Answer:** `LEAD(attendance)` returns the attendance value from the next row after sorting by attendance.

## Question 5

Assign students into 4 performance groups using NTILE().

```python
query = '''
SELECT
    student_id,
    name,
    score,
    NTILE(4) OVER (ORDER BY score DESC) AS performance_group
FROM Student_Performance;
'''
print(query)
```

**Answer:** `NTILE(4)` divides the students into four groups according to score, with the highest scores placed first.

## Question 6

For each course, assign a row number based on attendance (highest first).

```python
query = '''
SELECT
    student_id,
    name,
    course,
    attendance,
    ROW_NUMBER() OVER (
        PARTITION BY course
        ORDER BY attendance DESC
    ) AS attendance_row_number
FROM Student_Performance;
'''
print(query)
```

**Answer:** `PARTITION BY course` restarts the numbering for each course, and the highest attendance gets row number 1.

## Question 7

Calculate the number of days each student has been enrolled (from join_date to today). Assume current date = '2025-01-01'.

```python
query = '''
SELECT
    student_id,
    name,
    join_date,
    DATEDIFF('2025-01-01', join_date) AS days_enrolled
FROM Student_Performance;
'''
print(query)
```

**Answer:** `DATEDIFF()` calculates the number of days between each student's `join_date` and the given current date, `2025-01-01`.

## Question 8

Format join_date as “Month Year” (e.g., “June 2023”).

```python
query = '''
SELECT
    student_id,
    name,
    DATE_FORMAT(join_date, '%M %Y') AS formatted_join_date
FROM Student_Performance;
'''
print(query)
```

**Answer:** `DATE_FORMAT(join_date, '%M %Y')` displays the full month name followed by the four-digit year.

## Question 9

Replace the city ‘Mumbai’ with ‘MUM’ for display purposes.

```python
query = '''
SELECT
    student_id,
    name,
    REPLACE(city, 'Mumbai', 'MUM') AS displayed_city
FROM Student_Performance;
'''
print(query)
```

**Answer:** `REPLACE()` changes `Mumbai` to `MUM` only in the query output; it does not update the stored data.

## Question 10

For each course, find the highest score using FIRST_VALUE().

```python
query = '''
SELECT
    student_id,
    name,
    course,
    score,
    FIRST_VALUE(score) OVER (
        PARTITION BY course
        ORDER BY score DESC
    ) AS highest_score_in_course
FROM Student_Performance;
'''
print(query)
```

**Answer:** `FIRST_VALUE(score)` returns the first score in each course after sorting scores in descending order, which is the highest score for that course.

## Question 1

Define Covariance and explain how it differs from Correlation in terms of scale and interpretation.

**Answer:** Covariance measures how two variables change together.

- A **positive covariance** means the variables tend to increase together.
- A **negative covariance** means one variable tends to increase when the other decreases.
- A covariance near **zero** suggests little linear co-movement.

Covariance depends on the units and scale of the variables, so its magnitude is difficult to compare across datasets. Correlation is a standardized form of covariance and always lies between **-1 and +1**, making its strength and direction easier to interpret.

## Question 2

What does a positive, negative, and zero covariance indicate about the relationship between two variables?

**Answer:**  
- **Positive covariance:** The two variables generally move in the same direction.  
- **Negative covariance:** The variables generally move in opposite directions.  
- **Zero covariance:** There is little or no linear relationship between the variables.

## Question 3

Discuss the limitations of covariance as a measure of relationship between two variables. Why is correlation preferred in many cases?

**Answer:** Covariance has several limitations:

- Its value depends on the measurement units of the variables.
- Its magnitude has no fixed scale, so it is hard to judge whether a relationship is weak or strong.
- It is difficult to compare covariance values from different datasets.

Correlation is often preferred because it standardizes covariance into a value between **-1 and +1**, making both the direction and strength of a linear relationship easier to interpret and compare.

## Question 4

Explain the difference between Pearson’s correlation coefficient and Spearman’s rank correlation coefficient. When would you prefer to use Spearman’s correlation?

**Answer:** **Pearson’s correlation coefficient** measures the strength and direction of a **linear relationship** between two numerical variables.

**Spearman’s rank correlation coefficient** measures the strength and direction of a **monotonic relationship** using the ranks of the data rather than the raw values.

Spearman’s correlation is preferred when:

- The data is ordinal or ranked.
- The relationship is monotonic but not necessarily linear.
- The data contains outliers that may strongly affect Pearson correlation.
- The assumptions required for Pearson correlation are not appropriate.

## Question 5

If the correlation coefficient between two variables X and Y is 0.85, interpret this value in context. Can you infer causation from this value? Why or why not?

**Answer:** A correlation coefficient of **0.85** indicates a **strong positive linear relationship** between X and Y. As X increases, Y generally tends to increase as well.

However, correlation does **not** prove causation. The relationship may be influenced by other variables, coincidence, or a common underlying factor. Therefore, we cannot conclude that changes in X directly cause changes in Y based only on the correlation coefficient.

## Question 6

Using the dataset below, calculate the covariance between X and Y.

| X | 2 | 4 | 6 | 8 |
|---|---:|---:|---:|---:|
| Y | 3 | 7 | 5 | 10 |

```python
import numpy as np

X = np.array([2, 4, 6, 8], dtype=float)
Y = np.array([3, 7, 5, 10], dtype=float)

population_covariance = np.cov(X, Y, ddof=0)[0, 1]
sample_covariance = np.cov(X, Y, ddof=1)[0, 1]

print("Population covariance =", population_covariance)
print("Sample covariance =", sample_covariance)
```

**Answer:**  
The mean of X is **5** and the mean of Y is **6.25**.

Using the population covariance formula:

\[
\operatorname{Cov}(X,Y)=rac{\sum (x_i-ar{x})(y_i-ar{y})}{n}
\]

the covariance is **5.75**.

If the data is treated as a sample, the sample covariance is approximately **7.67**.

The positive covariance shows that X and Y generally move in the same direction.

## Question 7

Compute the Pearson correlation coefficient between variables A and B:

| A | 10 | 20 | 30 | 40 | 50 |
|---|---:|---:|---:|---:|---:|
| B | 8 | 14 | 18 | 24 | 28 |

```python
A = np.array([10, 20, 30, 40, 50], dtype=float)
B = np.array([8, 14, 18, 24, 28], dtype=float)

correlation = np.corrcoef(A, B)[0, 1]
print("Pearson correlation coefficient =", round(correlation, 4))
```

**Answer:** The Pearson correlation coefficient is approximately **0.998**.

This indicates an **extremely strong positive linear relationship** between A and B. As A increases, B also increases in a nearly linear pattern.

## Question 8

The following table shows heights (in cm) and weights (in kg) of 5 students.

| Height | 150 | 160 | 165 | 170 | 180 |
|---|---:|---:|---:|---:|---:|
| Weight | 50 | 55 | 58 | 62 | 70 |

Find the correlation coefficient between Height and Weight.

```python
height = np.array([150, 160, 165, 170, 180], dtype=float)
weight = np.array([50, 55, 58, 62, 70], dtype=float)

correlation = np.corrcoef(height, weight)[0, 1]
print("Correlation coefficient =", round(correlation, 4))
```

**Answer:** The correlation coefficient is approximately **0.990**.

This indicates a **very strong positive correlation** between height and weight in this dataset. In general, the taller students also have higher weights.

## Question 10

Two investment portfolios have the following returns (%) over 5 years. Compute the covariance and correlation coefficient, and interpret whether the portfolios move together.

| Year | Portfolio A | Portfolio B |
|---|---:|---:|
| 1 | 8 | 6 |
| 2 | 10 | 9 |
| 3 | 12 | 11 |
| 4 | 9 | 8 |
| 5 | 11 | 10 |

```python
portfolio_A = np.array([8, 10, 12, 9, 11], dtype=float)
portfolio_B = np.array([6, 9, 11, 8, 10], dtype=float)

population_cov = np.cov(portfolio_A, portfolio_B, ddof=0)[0, 1]
sample_cov = np.cov(portfolio_A, portfolio_B, ddof=1)[0, 1]
correlation = np.corrcoef(portfolio_A, portfolio_B)[0, 1]

print("Population covariance =", round(population_cov, 4))
print("Sample covariance =", round(sample_cov, 4))
print("Correlation coefficient =", round(correlation, 4))
```

**Answer:**  
The **population covariance** is approximately **2.80**, while the **sample covariance** is approximately **3.50**.

The Pearson correlation coefficient is approximately **0.986**.

Because both covariance and correlation are strongly positive, the two portfolios generally **move together in the same direction**. When Portfolio A has higher returns, Portfolio B also tends to have higher returns.

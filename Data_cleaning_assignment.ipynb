# Data Cleaning & Essential Functions Assignment

## Question
1. What is data cleaning, and why is it important in data analysis?

## Answer
Data cleaning is the process of identifying and correcting inaccurate, incomplete, duplicate, inconsistent, or improperly formatted data.

It is important because clean data improves the accuracy and reliability of analysis and helps in making better decisions.

## Question
1(a). What are the potential consequences of analyzing unclean or messy data?

## Answer
Analyzing unclean or messy data can lead to:
- Incorrect results
- Misleading conclusions
- Poor business decisions
- Duplicate or inconsistent records
- Errors in reports and calculations
- Wastage of time and resources

## Question
1(b). Explain the common steps involved in cleaning and organizing data.

## Answer
Common steps are:
1. Identify missing values.
2. Remove duplicate records.
3. Correct spelling and formatting errors.
4. Standardize text, date, and number formats.
5. Handle blank or invalid values.
6. Check for outliers or unusual values.
7. Organize columns and rows properly.
8. Validate the cleaned data before analysis.

## Question
3. Explain the use of text functions such as TRIM, LEFT, RIGHT, MID, and CONCAT in data cleaning.

## Answer
- **TRIM:** Removes extra spaces from text.  
  Example: `=TRIM(A1)`
- **LEFT:** Extracts characters from the beginning of a text string.  
  Example: `=LEFT(A1,5)`
- **RIGHT:** Extracts characters from the end of a text string.  
  Example: `=RIGHT(A1,4)`
- **MID:** Extracts characters from the middle of a text string.  
  Example: `=MID(A1,6,4)`
- **CONCAT:** Combines text from two or more cells/strings.  
  Example: `=CONCAT(A1," ",B1)`

These functions help clean, split, standardize, and combine text data.

## Question
4. What is the role of date functions like TODAY in managing datasets?

## Answer
The `TODAY()` function returns the current date automatically.

It is useful for:
- Calculating age
- Finding due dates
- Tracking deadlines
- Calculating the number of days remaining
- Creating reports that update with the current date

Formula:
`=TODAY()`

## Question
8(a). What is the difference between TODAY() and NOW() in Excel? Give an example of when to use each.

## Answer
- `TODAY()` returns the **current date only**.
- `NOW()` returns the **current date and current time**.

Examples:
- Use `=TODAY()` to calculate the number of days until a due date.
- Use `=NOW()` when you need a date-and-time timestamp.

## Question
8(b). If cell A1 contains the date 2025-06-10, write a formula using TODAY() to determine how many days are left until that date.

## Answer
Formula:

`=A1-TODAY()`

This returns the number of days remaining until the date in A1.

## Question
8(c). Write an Excel formula using NOW() to display the current date and time in the format MM/DD/YYYY HH:MM AM/PM.

## Answer
Formula:

`=TEXT(NOW(),"mm/dd/yyyy hh:mm AM/PM")`

## Question
8(d). If a cell contains =TODAY(), what will happen when the worksheet is reopened the next day?

## Answer
The value will automatically update to the new current date because `TODAY()` is a dynamic function that recalculates when the worksheet is opened or recalculated.

## Question
8(e). You want to store a static date (today’s date) in a cell without it changing every day. What keyboard shortcut should you use?

## Answer
Use:

**Ctrl + ;**

This inserts the current date as a static value, so it will not change automatically.

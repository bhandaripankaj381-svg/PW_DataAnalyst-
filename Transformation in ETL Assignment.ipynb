## Question 1

Define Data Transformation in ETL and explain why it is important.

**Answer:** Data Transformation is the process of converting extracted data into a clean, consistent, and usable format before loading it into the target system.

It is important because source data may contain different formats, missing values, duplicate records, inconsistent labels, or incompatible data types. Transformation improves data quality and ensures that the data can be correctly analyzed and stored in the destination system.

## Question 2

List any four common activities involved in Data Cleaning.

**Answer:** Four common Data Cleaning activities are:

1. **Handling missing values**
2. **Removing duplicate records**
3. **Correcting inconsistent or invalid values**
4. **Standardizing data formats**, such as dates, text case, and category names

## Question 3

What is the difference between Normalization and Standardization?

**Answer:**  

**Normalization** rescales values into a fixed range, usually **0 to 1**.

Formula:

\[
x' = \frac{x - x_{min}}{x_{max} - x_{min}}
\]

**Standardization** transforms values so that the data has a mean of **0** and a standard deviation of **1**.

Formula:

\[
z = \frac{x - \mu}{\sigma}
\]

Normalization is useful when a fixed range is required, while standardization is useful when variables need to be compared on the same scale.

## Question 4

A dataset has missing values in the “Age” column. Suggest two techniques to handle this and explain when they should be used.

**Answer:** Two techniques are:

1. **Mean or Median Imputation:** Replace missing Age values with the mean or median age. The median is often preferred when the data contains extreme values or is skewed.

2. **Remove Rows with Missing Age:** Delete records where Age is missing. This should be used when only a very small number of rows are missing and removing them will not significantly reduce the dataset or introduce bias.

## Question 5

Convert the following inconsistent “Gender” entries into a standardized format (“Male”, “Female”): `['M', 'male', 'F', 'Female', 'MALE', 'f']`

## Question 6

What is One-Hot Encoding? Give an example with the categories: “Red, Blue, Green”.

**Answer:** One-Hot Encoding converts categorical values into separate binary columns. Each category gets its own column, where `1` represents the presence of that category and `0` represents its absence.

Example:

| Color | Red | Blue | Green |
|---|---:|---:|---:|
| Red | 1 | 0 | 0 |
| Blue | 0 | 1 | 0 |
| Green | 0 | 0 | 1 |

This allows categorical data to be used in numerical analysis and machine learning models.

## Question 7

Explain the difference between Data Integration and Data Mapping in ETL.

**Answer:**  

**Data Integration** is the process of combining data from multiple sources into a unified dataset or target system.

**Data Mapping** defines how fields from the source correspond to fields in the target. For example, a source column named `cust_name` may be mapped to a target column named `CustomerName`.

In short, integration combines data, while mapping defines how source fields are connected to target fields during the ETL process.

## Question 8

Explain why Z-score Standardization is preferred over Min-Max Scaling when outliers exist.

**Answer:** Z-score Standardization is often preferred because Min-Max Scaling depends directly on the minimum and maximum values. An extreme outlier can stretch the range and compress most normal values into a very small interval.

Z-score Standardization uses the mean and standard deviation instead:

\[
z = \frac{x - \mu}{\sigma}
\]

It does not force all values into a fixed range, so the data is generally less compressed by extreme minimum or maximum values. However, strong outliers can still affect the mean and standard deviation.

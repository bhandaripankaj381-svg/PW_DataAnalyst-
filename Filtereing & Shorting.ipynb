## Question 1

Show employees working in either the ‘IT’ or ‘HR’ departments.

```python
query = '''
SELECT *
FROM Employees
WHERE Department IN ('IT', 'HR');
'''
print(query)
```

**Answer:** This query returns all employees whose department is either **IT** or **HR**.

## Question 2

Retrieve employees whose department is in ‘Sales’, ‘IT’, or ‘Finance’.

```python
query = '''
SELECT *
FROM Employees
WHERE Department IN ('Sales', 'IT', 'Finance');
'''
print(query)
```

**Answer:** This query filters the Employees table and returns employees working in **Sales, IT, or Finance**.

## Question 3

Display employees whose salary is between ₹50,000 and ₹70,000.

```python
query = '''
SELECT *
FROM Employees
WHERE Salary BETWEEN 50000 AND 70000;
'''
print(query)
```

**Answer:** This query returns employees whose salary is between **₹50,000 and ₹70,000**, including both limits.

## Question 4

List employees whose names start with the letter ‘A’.

```python
query = '''
SELECT *
FROM Employees
WHERE EmpName LIKE 'A%';
'''
print(query)
```

**Answer:** `LIKE 'A%'` selects employee names that begin with the letter **A**.

## Question 5

Find employees whose names contain the substring ‘an’.

```python
query = '''
SELECT *
FROM Employees
WHERE EmpName LIKE '%an%';
'''
print(query)
```

**Answer:** `LIKE '%an%'` finds employee names containing **an** anywhere in the name.

## Question 6

Show employees who are from ‘Delhi’ or ‘Mumbai’ and earn more than ₹55,000.

```python
query = '''
SELECT *
FROM Employees
WHERE City IN ('Delhi', 'Mumbai')
  AND Salary > 55000;
'''
print(query)
```

**Answer:** This query first selects employees from **Delhi or Mumbai**, then keeps only those earning more than **₹55,000**.

## Question 7

Display all employees except those from the ‘HR’ department.

```python
query = '''
SELECT *
FROM Employees
WHERE Department <> 'HR';
'''
print(query)
```

**Answer:** `<> 'HR'` excludes all employees whose department is **HR**.

## Question 8

Get all employees hired between 2019 and 2022, ordered by HireDate (oldest first).

```python
query = '''
SELECT *
FROM Employees
WHERE HireDate BETWEEN '2019-01-01' AND '2022-12-31'
ORDER BY HireDate ASC;
'''
print(query)
```

**Answer:** This query returns employees hired from **2019 through 2022** and sorts them by `HireDate` from the **oldest to the newest**.

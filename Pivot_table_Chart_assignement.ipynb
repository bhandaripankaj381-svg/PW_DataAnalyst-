# Pivot Table & Charts Assignment

## Question
1. What is the purpose of a PivotTable in Excel?

## Answer
A PivotTable is used to quickly summarize, analyze, compare, and organize large amounts of data.

It helps us:
- Calculate totals, averages, counts, etc.
- Compare sales by Region, Product, or Month.
- Rearrange data without changing the original dataset.
- Create reports and charts easily.

## Question
2. Create a PivotTable to show Total Sales by Region.

## Answer
**Steps:**
1. Select the complete dataset.
2. Go to **Insert → PivotTable**.
3. Choose **New Worksheet** and click **OK**.
4. Drag **Region** into the **Rows** area.
5. Drag **Sales** into the **Values** area.
6. Make sure the calculation is **Sum of Sales**.

The PivotTable will show total sales for:
- East
- West
- North
- South

**Note:** Exact totals depend on the actual Sales Dataset linked in the assignment.

## Question
3. Apply Conditional Formatting to highlight sales greater than 400.

## Answer
**Steps:**
1. Select the cells in the **Sales** column.
2. Go to **Home → Conditional Formatting**.
3. Select **Highlight Cells Rules → Greater Than**.
4. Enter `400`.
5. Choose a highlighting format.
6. Click **OK**.

All Sales values greater than 400 will be highlighted.

## Question
4. Explain the difference between Sorting and Filtering with an example from this dataset.

## Answer
**Sorting** changes the order of data.

Example:
- Sort the **Sales** column from Largest to Smallest.
- The rows will be rearranged according to Sales values.

**Filtering** temporarily shows only the rows that match a condition.

Example:
- Filter **Region = East**.
- Only records from the East region will be displayed.

So, sorting changes the order, while filtering hides unwanted records temporarily.

## Question
5. Create a PivotChart showing Monthly Sales Trend.

## Answer
**Steps:**
1. Create a PivotTable from the dataset.
2. Drag **Date** into the **Rows** area.
3. Drag **Sales** into the **Values** area.
4. Right-click any date in the PivotTable.
5. Choose **Group**.
6. Select **Months**.
7. Go to **PivotTable Analyze → PivotChart**.
8. Select a **Line Chart**.
9. Click **OK**.

The chart will show the monthly sales trend from January to June 2024.

## Question
6. Use a Slicer to filter data by Product and observe changes in sales.

## Answer
**Steps:**
1. Click anywhere inside the PivotTable.
2. Go to **PivotTable Analyze → Insert Slicer**.
3. Select **Product**.
4. Click **OK**.
5. Use the slicer buttons such as:
   - Apples
   - Oranges
   - Bananas
   - Mangoes

When a product is selected, the PivotTable and connected PivotChart will update automatically.

## Question
7. Why are Dashboards important for decision-making in business analysis?

## Answer
Dashboards are important because they present important business information in a simple visual format.

They help decision-makers:
- Track key performance indicators.
- Compare performance across regions and products.
- Identify trends and patterns.
- Find problems quickly.
- Make faster and better decisions.
- View multiple reports in one place.

## Question
9. Group dates in the PivotTable into Months and calculate the average sales per month.

## Answer
**Steps:**
1. Create a PivotTable.
2. Drag **Date** into the **Rows** area.
3. Drag **Sales** into the **Values** area.
4. Right-click any date.
5. Click **Group**.
6. Select **Months**.
7. In the Values area, click **Sum of Sales → Value Field Settings**.
8. Select **Average**.
9. Click **OK**.

The PivotTable will now display **Average Sales per Month**.

**Note:** Exact monthly averages require the actual Sales Dataset.

## Question
10. Combine Apples and Oranges into one group called Fruits and compare its sales with other products.

## Answer
**Steps:**
1. Create a PivotTable.
2. Put **Product** in the **Rows** area.
3. Put **Sales** in the **Values** area.

7. Rename the group to **Fruits**.
8. Keep Bananas and Mangoes as separate products.
9. Compare **Sum of Sales** for Fruits with the other products.

This allows Apples and Oranges to be analyzed together as one group.

**Note:** The exact comparison depends on the values in the linked Sales Dataset.

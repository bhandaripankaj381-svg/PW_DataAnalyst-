## Question 1

Describe different types of data sources used in ETL with suitable examples.

**Answer:** ETL systems use different data sources such as **relational databases** (MySQL, PostgreSQL, SQL Server), **flat files** (CSV, TXT), **Excel files**, **APIs**, **cloud storage**, and **NoSQL databases** such as MongoDB. An ETL pipeline can extract data from one or several of these sources before transformation and loading.

## Question 2

What is data extraction? Explain its role in the ETL pipeline.

**Answer:** Data extraction is the process of retrieving required data from source systems such as databases, files, or APIs. It is the first stage of **ETL (Extract, Transform, Load)**. The extracted data is passed to a staging or processing area so it can be cleaned, transformed, validated, and loaded into the target system.

## Question 3

Explain the difference between CSV and Excel in terms of extraction and ETL usage.

**Answer:** **CSV** is a lightweight plain-text format that normally stores one table and is easy and fast to process in automated ETL. **Excel** files can contain multiple worksheets, formulas, formatting, and other spreadsheet features, so they require an Excel-compatible connector or library. CSV is generally simpler for large automated transfers, while Excel is common for smaller business-maintained datasets.

## Question 4

Explain the steps involved in extracting data from a relational database.

**Answer:** First identify the database, tables, and required columns. Establish a database connection, write and execute a SQL query, and apply filters to retrieve only required records. Then validate the extracted data for completeness and correct data types, handle errors, and send the results to a staging area for transformation. For recurring jobs, incremental extraction can retrieve only new or changed records.

## Question 5

Explain three common challenges faced during data extraction.

**Answer:** Three common challenges are **data quality problems** such as missing or duplicate values, **different source formats and structures** across databases/files/APIs, and **large data volumes** that can cause memory and performance problems. Validation, standard extraction logic, filtering, batching, and chunk processing can help address these issues.

## Question 6

What are APIs? Explain how APIs help in real-time data extraction.

**Answer:** An **API (Application Programming Interface)** allows applications to request and exchange data using defined endpoints. ETL systems can call APIs to obtain current data, commonly returned as JSON or XML. This supports frequent or near-real-time extraction. ETL developers should also handle authentication, pagination, rate limits, and API errors.

## Question 7

Why are databases preferred for enterprise-level data extraction?

**Answer:** Databases provide structured storage, efficient SQL querying, scalability, security, data integrity, and controlled concurrent access. They can manage large datasets and allow ETL systems to filter records and perform incremental extraction. These features make databases reliable sources for enterprise ETL pipelines.

## Question 8

What steps should an ETL developer take when extracting data from large CSV files (1GB+)?

**Answer:** First inspect the file structure, delimiter, encoding, columns, and data types. Do not load the whole file into memory; instead, read it in **chunks or batches**. Select only required columns, validate each chunk, process and load chunks progressively, monitor resource usage, and use error handling and logging. Chunk processing makes extraction of 1GB+ CSV files more memory-efficient.

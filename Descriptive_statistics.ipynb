## Q1: Understanding Central Tendency (Easy)

A bakery tracks the daily sales of muffins (in dozens) over a week: `[10, 12, 11, 15, 14, 13, 12]`.

**What is the most representative value of their weekly sales, and why?**

```python
sales = [10, 12, 11, 15, 14, 13, 12]

mean_sales = np.mean(sales)
median_sales = np.median(sales)
mode_sales = multimode(sales)

print('Mean:', round(mean_sales, 2))
print('Median:', median_sales)
print('Mode:', mode_sales)
```

**Answer:** A good representative value is **12 dozen muffins**. The median is 12 and the mode is also 12, while the mean is about 12.43. Since the values are fairly close together and 12 is both the middle and most frequent value, it represents the week's daily sales well.

## Q4: Median in Real Life (Medium)

A car dealer notes the prices of used cars: `[$8,000, $9,500, $10,200, $11,000, $50,000]`.

**Why is the median a better measure than the mean in this case? Calculate the median.**

```python
car_prices = [8000, 9500, 10200, 11000, 50000]
median_price = np.median(car_prices)
mean_price = np.mean(car_prices)

print('Median price: $', format(median_price, ',.0f'), sep='')
print('Mean price: $', format(mean_price, ',.0f'), sep='')
```

**Answer:** The median price is **$10,200**. The $50,000 car is much more expensive than the others and pulls the mean upward. The median is less affected by this extreme value, so it better represents a typical used-car price.

## Q5: Dispersion Introduction (Medium)

A student times how long it takes to finish a puzzle each day: `[25, 30, 27, 35, 40]`.

**What does the range tell us about the variation in the student’s puzzle-solving time?**

```python
puzzle_times = [25, 30, 27, 35, 40]
puzzle_range = max(puzzle_times) - min(puzzle_times)
print('Range:', puzzle_range, 'minutes')
```

**Answer:** The range is **15 minutes**. The student's fastest time is 25 minutes and slowest time is 40 minutes, so the completion times vary across a 15-minute interval.

## Q7: Variance for Decision-Making (Medium)

Two delivery companies track delivery delays (in minutes).

- Company A: variance = 6
- Company B: variance = 15

**Which company is more consistent, and why?**

```python
variance_A = 6
variance_B = 15

more_consistent = 'Company A' if variance_A < variance_B else 'Company B'
print('More consistent:', more_consistent)
```

**Answer:** **Company A** is more consistent because its variance (**6**) is lower than Company B's (**15**). A lower variance means the delivery delays are more closely grouped and therefore less variable.

## Q8: Standard Deviation in Context (Hard)

A finance student compares the daily price fluctuations of two cryptocurrencies.

## Q9: Combining Measures (Hard)

A family records their monthly electricity usage (in kWh): `[400, 420, 390, 450, 410]`.

**Find the mean and standard deviation. What do these values together tell you about the family’s energy use pattern?**

```python
electricity = [400, 420, 390, 450, 410]
mean_electricity = np.mean(electricity)
std_electricity = np.std(electricity, ddof=0)

print('Mean:', round(mean_electricity, 2), 'kWh')
print('Population standard deviation:', round(std_electricity, 2), 'kWh')
```

**Answer:** The mean electricity usage is **414 kWh**, and the population standard deviation is approximately **20.59 kWh**. This means the family's monthly usage is centered around 414 kWh and usually does not vary very widely from that average, indicating a fairly consistent energy-use pattern.

## Q10: Practical Application (Hard)

A basketball player’s points in 8 games are recorded: `[15, 18, 20, 22, 25, 17, 19, 21]`.

**Find the mean, median, mode, range, and standard deviation. What insights can these measures provide about the player’s scoring performance?**

```python
points = [15, 18, 20, 22, 25, 17, 19, 21]

mean_points = np.mean(points)
median_points = np.median(points)
modes = multimode(points)
has_mode = len(modes) < len(set(points))
points_range = max(points) - min(points)
std_points = np.std(points, ddof=0)

print('Mean:', round(mean_points, 3))
print('Median:', median_points)
print('Mode:', modes if has_mode else 'No mode (all scores occur once)')
print('Range:', points_range)
print('Population standard deviation:', round(std_points, 3))
```

**Answer:**

- **Mean:** 19.625 points
- **Median:** 19.5 points
- **Mode:** No mode, because every score occurs once
- **Range:** 10 points
- **Population standard deviation:** approximately 2.956 points

These measures show that the player typically scores around **19–20 points per game**. The relatively small standard deviation suggests the scoring is fairly consistent, while the 10-point range shows the total difference between the lowest score (15) and highest score (25).

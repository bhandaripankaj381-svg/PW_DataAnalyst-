## Question 1

Count how many cities are there in each country?

```python
query = '''
USE world;

SELECT
    c.Name AS Country,
    COUNT(ci.ID) AS CityCount
FROM country c
LEFT JOIN city ci
    ON c.Code = ci.CountryCode
GROUP BY c.Code, c.Name
ORDER BY CityCount DESC;
'''
print(query)
```

**Answer:** This groups cities by country and counts how many cities belong to each country.

## Question 2

Display all continents having more than 30 countries.

```python
query = '''
USE world;

SELECT
    Continent,
    COUNT(*) AS CountryCount
FROM country
GROUP BY Continent
HAVING COUNT(*) > 30;
'''
print(query)
```

**Answer:** `GROUP BY` creates one row per continent, and `HAVING` keeps only continents with more than 30 countries.

## Question 3

List regions whose total population exceeds 200 million.

```python
query = '''
USE world;

SELECT
    Region,
    SUM(Population) AS TotalPopulation
FROM country
GROUP BY Region
HAVING SUM(Population) > 200000000
ORDER BY TotalPopulation DESC;
'''
print(query)
```

**Answer:** This sums the population of all countries in each region and keeps regions whose total exceeds 200 million.

## Question 4

Find the top 5 continents by average GNP per country.

```python
query = '''
USE world;

SELECT
    Continent,
    AVG(GNP) AS AverageGNP
FROM country
GROUP BY Continent
ORDER BY AverageGNP DESC
LIMIT 5;
'''
print(query)
```

**Answer:** This calculates average GNP for each continent, sorts from highest to lowest, and returns the top five.

## Question 5

Find the total number of official languages spoken in each continent.

```python
query = '''
USE world;

SELECT
    c.Continent,
    COUNT(DISTINCT cl.Language) AS OfficialLanguageCount
FROM country c
JOIN countrylanguage cl
    ON c.Code = cl.CountryCode
WHERE cl.IsOfficial = 'T'
GROUP BY c.Continent
ORDER BY OfficialLanguageCount DESC;
'''
print(query)
```

**Answer:** This joins countries with their languages, keeps only official languages, and counts distinct official languages per continent.

## Question 6

Find the maximum and minimum GNP for each continent.

```python
query = '''
USE world;

SELECT
    Continent,
    MAX(GNP) AS MaximumGNP,
    MIN(GNP) AS MinimumGNP
FROM country
GROUP BY Continent;
'''
print(query)
```

**Answer:** This uses `MAX()` and `MIN()` to find the highest and lowest GNP values within each continent.

## Question 7

Find the country with the highest average city population.

```python
query = '''
USE world;

SELECT
    c.Name AS Country,
    AVG(ci.Population) AS AverageCityPopulation
FROM country c
JOIN city ci
    ON c.Code = ci.CountryCode
GROUP BY c.Code, c.Name
ORDER BY AverageCityPopulation DESC
LIMIT 1;
'''
print(query)
```

**Answer:** This computes the average population of cities for each country and returns the country with the highest average.

## Question 8

List continents where the average city population is greater than 200,000.

```python
query = '''
USE world;

SELECT
    c.Continent,
    AVG(ci.Population) AS AverageCityPopulation
FROM country c
JOIN city ci
    ON c.Code = ci.CountryCode
GROUP BY c.Continent
HAVING AVG(ci.Population) > 200000
ORDER BY AverageCityPopulation DESC;
'''
print(query)
```

**Answer:** This calculates average city population by continent and filters to averages above 200,000.

## Question 9

Find the total population and average life expectancy for each continent, ordered by average life expectancy descending.

```python
query = '''
USE world;

SELECT
    Continent,
    SUM(Population) AS TotalPopulation,
    AVG(LifeExpectancy) AS AverageLifeExpectancy
FROM country
GROUP BY Continent
ORDER BY AverageLifeExpectancy DESC;
'''
print(query)
```

**Answer:** This returns total population and average life expectancy for each continent, sorted by highest average life expectancy first.

## Question 10

Find the top 3 continents with the highest average life expectancy, but only include those where the total population is over 200 million.

```python
query = '''
USE world;

SELECT
    Continent,
    SUM(Population) AS TotalPopulation,
    AVG(LifeExpectancy) AS AverageLifeExpectancy
FROM country
GROUP BY Continent
HAVING SUM(Population) > 200000000
ORDER BY AverageLifeExpectancy DESC
LIMIT 3;
'''
print(query)
```

**Answer:** This filters continents to those with total population above 200 million, sorts by average life expectancy, and returns the top three.

## Question 1

Simulate 30 rolls with `=RANDBETWEEN(1,6)`. What is the probability of rolling a 3 exactly 5 times? (Hint: Use `BINOM.DIST`)

```python
from math import comb
n, k, p = 30, 5, 1/6
probability = comb(n,k)*(p**k)*((1-p)**(n-k))
print(f"Probability = {probability:.6f}")
print(f"Percentage = {probability*100:.2f}%")
```

**Answer:** The probability is approximately **0.1921**, or **19.21%**.

Excel formula: `=BINOM.DIST(5,30,1/6,FALSE)`

## Question 2

Generate 100 values in Excel using the continuous uniform distribution `RAND()` and plot a histogram. Describe the shape of the distribution.

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
values = np.random.uniform(0,1,100)
plt.hist(values, bins=10, edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of 100 Uniform Random Values")
plt.show()
```

**Answer:** The histogram is approximately **uniform (flat/rectangular)** because values from 0 to 1 are equally likely. With 100 observations, some random variation between the bars is expected.

## Question 3

A dataset has a mean of 50 and a standard deviation of 5. What percentage of values lie between 45 and 55 if the data follows a normal distribution?

```python
from math import erf, sqrt
mu, sigma = 50, 5
cdf = lambda x: 0.5*(1+erf((x-mu)/(sigma*sqrt(2))))
probability = cdf(55)-cdf(45)
print(f"Percentage = {probability*100:.2f}%")
```

**Answer:** 45 and 55 correspond to z-scores of -1 and +1. Approximately **68.27%** of values in a normal distribution lie within one standard deviation of the mean.

## Question 4

What is the concept of standardization (z-score), and why is it important in data analysis? Explain the formula and how standardization transforms a dataset.

```python
data = np.array([40,45,50,55,60], dtype=float)
mu = np.mean(data)
sigma = np.std(data)
z = (data-mu)/sigma
print("Original:", data)
print("Z-scores:", np.round(z,3))
print("New mean:", round(np.mean(z),10))
print("New standard deviation:", np.std(z))
```

**Answer:** Standardization converts a value into a z-score using

\[
z=\frac{x-\mu}{\sigma}
\]

where \(x\) is the original value, \(\mu\) is the mean, and \(\sigma\) is the standard deviation. The transformed data has a mean of **0** and standard deviation of **1**. This makes variables measured on different scales easier to compare and use together in statistical analysis.

## Question 5

What is Kurtosis and their type?

**Answer:** Kurtosis describes a distribution's tails and tendency to produce extreme observations relative to a normal distribution.

1. **Mesokurtic:** Similar kurtosis to a normal distribution.
2. **Leptokurtic:** Heavier tails and greater tendency for extreme values.
3. **Platykurtic:** Lighter tails and fewer extreme values than a normal distribution.

## Question 6

Explain why the uniform distribution is a good model for the outcome of rolling a fair die.

```python
outcomes = range(1,7)
for x in outcomes:
    print(f"P({x}) = {1/6:.4f}")
```

**Answer:** A fair die has six possible outcomes, and every outcome has the same probability, **1/6**. Therefore, rolling a fair die follows a **discrete uniform distribution**.

## Question 7

Use Excel to compute the probability of getting at least 8 successes in 15 trials with success probability 0.5.

```python
n, p = 15, 0.5
probability = sum(comb(n,k)*(p**k)*((1-p)**(n-k)) for k in range(8,n+1))
print(f"Probability = {probability:.6f}")
print(f"Percentage = {probability*100:.2f}%")
```

**Answer:** The probability is **0.5**, or **50%**.

Excel formula: `=1-BINOM.DIST(7,15,0.5,TRUE)`

## Question 8

How does log transformation help in stabilizing variance and making data more normally distributed?

```python
data = np.array([1,2,3,5,10,20,50,100], dtype=float)
log_data = np.log(data)
print("Original:", data)
print("Log transformed:", np.round(log_data,3))
```

**Answer:** Log transformation compresses large values more strongly than small values. It can reduce right skewness, reduce the influence of very large observations, and stabilize variance when variability increases with the size of the observations. This can make some positively skewed datasets more symmetric and closer to a normal distribution.

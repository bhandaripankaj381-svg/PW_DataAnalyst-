## Question 1

What is a null hypothesis (H₀) and why is it important in hypothesis testing?

**Answer:** The **null hypothesis (H₀)** is a statement that there is no significant effect, difference, or relationship in the population. It is important because hypothesis testing begins by assuming H₀ is true and then uses sample evidence to decide whether there is enough evidence to reject it.

## Question 2

What does the significance level (α) represent in hypothesis testing?

**Answer:** The **significance level (α)** is the maximum probability of making a Type I error that the researcher is willing to accept. A common choice is α = 0.05, meaning a 5% risk of rejecting a true null hypothesis.

## Question 3

Differentiate between Type I and Type II errors.

**Answer:**  
- **Type I error:** Rejecting H₀ when H₀ is actually true. Its probability is α.  
- **Type II error:** Failing to reject H₀ when H₀ is actually false. Its probability is β.

## Question 4

Explain the difference between a one-tailed and two-tailed test. Give an example of each.

**Answer:**  
A **one-tailed test** checks for an effect in only one direction.

Example:  
H₀: μ = 50  
H₁: μ > 50

A **two-tailed test** checks for a difference in either direction.

Example:  
H₀: μ = 50  
H₁: μ ≠ 50

## Question 5

A company claims that the average time to resolve a customer complaint is 10 minutes.

A random sample of 9 complaints gives an average time of 12 minutes and a standard deviation of 3 minutes.

At α = 0.05, test the claim.

```python
import math
from scipy.stats import t

mu0 = 10
x_bar = 12
s = 3
n = 9
alpha = 0.05

t_stat = (x_bar - mu0) / (s / math.sqrt(n))
df = n - 1
p_value = 2 * (1 - t.cdf(abs(t_stat), df))

print("t-statistic =", round(t_stat, 4))
print("degrees of freedom =", df)
print("p-value =", round(p_value, 4))
```

**Answer:**  
H₀: μ = 10 minutes  
H₁: μ ≠ 10 minutes

The test statistic is **t = 2.00** with **8 degrees of freedom**.  
The two-tailed p-value is approximately **0.0805**.

Since **p > 0.05**, we **fail to reject H₀**.

There is not enough evidence at the 5% significance level to conclude that the true average complaint-resolution time is different from 10 minutes.

## Question 6

When should you use a Z-test instead of a t-test?

**Answer:** A **Z-test** is generally used when the population standard deviation is known or when the sample size is large enough for the normal approximation to be appropriate. A **t-test** is preferred when the population standard deviation is unknown, especially for small samples.

## Question 7

The productivity of 6 employees was measured before and after a training program.

| Employee | Before | After |
|---|---:|---:|
| 1 | 50 | 55 |
| 2 | 60 | 65 |
| 3 | 58 | 59 |
| 4 | 55 | 58 |
| 5 | 62 | 63 |
| 6 | 56 | 59 |

At α = 0.05, test if the training improved productivity.

```python
import numpy as np
from scipy.stats import ttest_rel

before = np.array([50, 60, 58, 55, 62, 56], dtype=float)
after  = np.array([55, 65, 59, 58, 63, 59], dtype=float)

differences = after - before
t_stat, p_two_tailed = ttest_rel(after, before)
p_one_tailed = p_two_tailed / 2 if t_stat > 0 else 1 - p_two_tailed / 2

print("Differences:", differences)
print("Mean difference =", round(np.mean(differences), 4))
print("t-statistic =", round(t_stat, 4))
print("one-tailed p-value =", round(p_one_tailed, 4))
```

**Answer:**  
Use a **paired t-test** because the same employees are measured before and after training.

H₀: The mean improvement is 0 or less.  
H₁: The mean improvement is greater than 0.

The mean improvement is **3 points**.  
The paired t-statistic is approximately **4.7434**, and the one-tailed p-value is approximately **0.0026**.

Since **p < 0.05**, we **reject H₀**.

There is significant evidence that the training **improved productivity**.

## Question 8

A company wants to test if product preference is independent of gender.

| Gender | Product A | Product B | Total |
|---|---:|---:|---:|
| Male | 30 | 20 | 50 |
| Female | 10 | 40 | 50 |
| Total | 40 | 60 | 100 |

At α = 0.05, test independence.

```python
import numpy as np
from scipy.stats import chi2_contingency

observed = np.array([
    [30, 20],
    [10, 40]
])

chi2, p_value, dof, expected = chi2_contingency(observed, correction=False)

print("Chi-square statistic =", round(chi2, 4))
print("degrees of freedom =", dof)
print("p-value =", round(p_value, 6))
print("Expected frequencies:")
print(expected)
```

**Answer:**  
H₀: Product preference is independent of gender.  
H₁: Product preference is associated with gender.

The expected frequencies are:

- Male, Product A: 20
- Male, Product B: 30
- Female, Product A: 20
- Female, Product B: 30

The chi-square statistic is approximately **16.6667** with **1 degree of freedom**.  
The p-value is approximately **0.000045**.

Since **p < 0.05**, we **reject H₀**.

There is significant evidence that **product preference is associated with gender**.
